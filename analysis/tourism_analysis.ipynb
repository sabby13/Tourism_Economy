# Tourism as an Economic Engine — Saudi Arabia Vision 2030
This notebook reconstructs the recoverable analytical workflow from the surviving project draft. Missing historical observations are intentionally not fabricated.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

DATA = Path('../data/recovered_observations.csv')
df = pd.read_csv(DATA)
df
```

## Recovered observations
The surviving draft explicitly reports selected GDP and tourism-expenditure observations. The 2021 tourism figure is described only as 'over 25 billion', so it is kept missing rather than converted into an invented exact value.

```python
fig, ax = plt.subplots(figsize=(8,4.5))
ax.plot(df['year'], df['gdp_usd_billion'], marker='o')
ax.set_title('Recovered Saudi Arabia GDP observations')
ax.set_xlabel('Year')
ax.set_ylabel('GDP (US$ billions)')
ax.grid(alpha=.25)
plt.tight_layout()
plt.show()
```

```python
tourism = df.dropna(subset=['tourism_expenditure_usd_billion'])
fig, ax = plt.subplots(figsize=(8,4.5))
ax.plot(tourism['year'], tourism['tourism_expenditure_usd_billion'], marker='o')
ax.set_title('Recovered international tourism expenditure observations')
ax.set_xlabel('Year')
ax.set_ylabel('Tourism expenditure (US$ billions)')
ax.grid(alpha=.25)
plt.tight_layout()
plt.show()
```

## ARIMA forecasting scaffold
The original project describes ARIMA forecasting toward 2030. The cell below is intentionally gated: ARIMA should be fitted only after a sufficiently complete annual series has been restored from the cited sources.

```python
def fit_arima_when_restored(series, order=(1,1,1), steps=7):
    from statsmodels.tsa.arima.model import ARIMA
    series = pd.Series(series).dropna()
    if len(series) < 10:
        raise ValueError('Restore at least 10 annual observations before fitting ARIMA.')
    model = ARIMA(series, order=order).fit()
    return model.forecast(steps=steps)

# Example after restoring the complete annual series:
# forecast = fit_arima_when_restored(full_df.set_index('year')['gdp_usd_billion'])
```
